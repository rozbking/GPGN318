## Objectives

Our objectives are:
* Design and perform a surface gravity survey to detect the underground steam tunnel in front of the Green Center.


## Background
### Tunnels

In previous labs you have modeled the tunnels, the approximate range of gravity anomalies was between 30-80 microGal. The tunnel is a void (or near-void) in the ground, so the anomaly should be **negative** directly above it.

This is a small signal. Wind, footsteps, poor leveling and instrument drift can easily add noise of the same size. How carefully we collect data decides whether we can see the tunnel at all.


## Task
Broadly your tasks involve three items:

### 1) Layout your survey
Setup and mark your survey locations on the ground.

### 2) Accurately measure the spatial locations
Using an appropriately accurate device, measure the station locations.

### 3) Measure Gravity
Using the gravimeters measure the anomaly of the tunnel. Remember to visit a base station often enough to accurately correct for instrument drift.

Afterwards, download the spatial and gravity data from the instruments.

## Before you go: think about these

Come to lab with an answer to each of these in your field notebook. There isn't one right answer; your report should explain what you chose and why.

1. **Station spacing and line length.** Using your Lab 5 model, how close together do stations need to be over the tunnel to see the anomaly, and how far out does the line need to go to see the background?
2. **Recording length.** How long should each reading last, and how many readings should you take at each station?
3. **Repeats.** Which stations should be measured more than once, and why? How will repeats tell you how noisy your data are?
4. **Base station.** Where should the base station be, and how often should we return to it? What happens to the anomaly shape if drift isn't corrected?
5. **Handoffs.** With one meter and five teams, how should the meter pass between teams so drift can still be corrected? (Hint: think about where each team starts and ends.)
6. **Best practices.** What can each operator do to get a good reading (ground, leveling, meter orientation, wind, sun, carrying the meter)?
7. **Time.** We have roughly 2 hours of measuring time. How many station readings, including repeats and base returns, can we realistically fit in?
