## Proton-precession
<div class="row">
<div class="column">

How it works:
* Electrically generate a strong magnetic field to orient a hydrogen rich fluid (Hydrogen atoms = protons).
* Release the magnetic field, the protons start to precess (like spinning a top)
* The rate of precession is linearly related to the strength of the magnetic field (thus a scalar magnetometer).
</div>
<div class="column"></div>

![Procession](https://upload.wikimedia.org/wikipedia/commons/8/82/Gyroscope_precession.gif?20070105133013)

## Optically pumped Alkali Vapor magnetometers


How it works:
* Uses an alkaline vapor (caesium, potassium, etc.)
* Sends photons through the vapor (laser)
    * alkaline vapor absorbs photon
    * excites the electrons to a higher energy
    * then emits a photon
    * electrons fall to a lower energy level
    * the vapor is considered transparent then (photon in -> photon out)
* Magnetic field strength change this process by making it less transparent
    * Scalar magnetometer

## Fluxgates
<div class="row">
<div class="column">


* Two coils wound around magnetically susceptible cores
* The materials are driven to magnetic saturation by an alternating current
    * magnetic saturation, means increasing the inducing field strength no longer increases the magnetic dipole moment.
* In the presence of an external field, the coils are more easily saturated in the direction of the external field.
* Measure the magnetic field in a single direction (So use multiple to get multiple components)

### Processing: Regional/Background Removal
* Any magnetic measurement is superposition of fields from many objects at different scales
* Example: magnetic data for UXO could include

* Background field is generally anything that is smoothly varying over your region of interest and is much larger than the footprint of the body you are interested in.
* Deciding what is background is a **subjective decision**.

### Processing: Reduction to Pole (RTP)
Same object buried at different locations on the earth yields different total field anomalies
* Can we undo that?

  

Filter the data to emulate the response as if the survey was taken at the pole. (Earth’s field is vertical; measure vertical component of the anomalous field)



This simplifies interpretation. Causative body lies beneath the peak

* Magnetic signatures are complicated because the same object provides different data depending where it on the earth (ie depends upon strength and direction of the inducing field.
* This ambiguity can be reduced by processing the data by “Reducing to the Pole”
* Unstable at low latitudes (horizontal inducing field)
