### Magnetic materials and magnetization
* Mineral grains contain magnetic domains that behave like small bar magnets
* Bar magnets have a North and a South pole $\rightarrow$ magnetic dipole
* Each magnetic domain (dipole) contributes towards the overall magnetization of a material

### Dipole moment and magnetization
* Magnetic dipole moment $\vec{m}_i$ defines the strength and orientation of a magnetization (domain or rock)  
* Magnetization is average magnetic moment per unit volume
  $$ \vec{M} = \frac{\sum \vec{m}_i}{V}$$
* Units: dipole moment per unit volume Ampere/meter (A/m)



* Integrate magnetism to get dipole moment for a whole sample

$$ m = \int_V M d V $$

### Sources of magnetism
* Magnetism results from movement of electric charges
* A current loop produces magnetic fields
* Atoms can have a dipole moment and produce fields:
    * Electrons and protons spin on their axis
    * Electrons circulate around the atomic nucleus.
* A magnetic domain is a region of uniform atomic spin


## Physical Properties


Fundamentally:
* The magnetization changes when a magnetic field is applied
* **Induced magnetization** lies along direction of the applied field

* How do we define the induced magnetization due to an applied magnetic field?

### Magnetic Susceptibility
Defines strength of magnetization due to an external field
$$\vec{M} = \frac{\sum \vec{m}_i}{V}$$
$$\vec{M} = \kappa \vec{H}$$

* Magnetization parallel to inducing field
* Susceptibility units: dimensionless

### Different mechanisms for induced magnetization
$$\vec{M} = \kappa \vec{H}$$
* Ferromagnetic Materials: $\kappa \approx +10 \rightarrow >1000$
* Ferrimagnetic Materials: $\kappa \approx +1 \rightarrow +10$
* Antiferromagnetic Materials: $\kappa \approx +10^{-4} \rightarrow +10^{-3}$ 
* Paramagnetic Materials: $\kappa \approx +10^{-5} \rightarrow +10^{-2}$
* Diamagnetic Materials: $\kappa \approx -10^{-5}$.

### Diamagnetic Materials
$\kappa \approx -10^{-5}$
* Oppose the applied magnetic field because that field changes the magnetic moments of electrons in their orbitals in a systematic way
* All materials are technically diamagnetic, but other forms of magnetization (if present) are much stronger
* Water, halite, graphite, calcite and quartz are common diamagnetic minerals
* Usually has all paired electrons



### Paramagnetic Materials
$\kappa \approx +10^{-5} \rightarrow +10^{-2}$
* Have unpaired electrons which align in a magnetic field to create an induced magnetic field
* Many metals are paramagnetic.



## Magnetic Fields and Fluxes
### and permeability

We have $$\vec{M} = \kappa \vec{H}$$
and
* $\vec{B}$: Magnetic Flux Density $\frac{Wb (\text{Weber})}{m^2} = T (\text{Tesla})$
* $\vec{H}$: Magnetic Field $\frac{A}{m}$
* $\vec{B} = \mu \vec{H}$

* $\mu$: Magnetic Permeability $\frac{H (\text{Henries})}{m}$
* $\mu = \mu_0 (1 + \kappa)$
* $\mu_0 = 4\pi \times 10^{-7}$ H/m

* $\vec{B} = \mu \vec{H} = \mu_0(1 + \kappa) \vec{H} = \mu_0(\vec{H} + \vec{M})$
* $\vec{H}\rightarrow$ **Inducing** magnetic field
* $\vec{M}\rightarrow$ **Induced** magnetism

## Recap
* Earth materials contain magnetic domains that behave like magnetic dipoles
* Magnetization is the total dipole moment per unit volume
* Magnetic dipoles will re-orient along the direction of an inducing field
* The strength of induced magnetization is defined by the magnetic susceptibility
* Magnetization produces a magnetic field
* Field lines go from North to South pole
* Magnetic fields and fluxes are related by the magnetic permeability

## Remanent Magnetization
<div class="row">
<div class="column">

* A permanent magnetization contribution which is not supported by an external field
* Magma cools below Curie temperature$\rightarrow$ Magnetic dipoles align along Earth’s field and leave permanent imprint
* Total magnetization is vector sum:
  $$ \vec{M}_T = \vec{M}_I + \vec{M}_R $$
* Only significant in ferro/ferrimagnetic materials!!! (magnetite, steel etc…)
</div>
<div class="column">


</div>
</div>
