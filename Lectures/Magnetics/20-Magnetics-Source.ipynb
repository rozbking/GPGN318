## Magnetic Fields and Fluxes
### and permeability
* $\vec{B}$: Magnetic Flux Density $\frac{Wb (\text{Weber})}{m^2} = T (\text{Tesla})$
* $\vec{H}$: Magnetic Field $\frac{A}{m}$
* $\vec{B} = \mu \vec{H}$

* $\mu$: Magnetic Permeability $\frac{H (\text{Henries})}{m}$
* $\mu = \mu_0 (1 + \kappa)$
* $\mu_0 = 4\pi \times 10^{-7}$ H/m

* Recall $\vec{M} = \kappa \vec{H}$ (Magnetic Susceptibility slide), so:
* $\vec{B} = \mu \vec{H} = \mu_0(1 + \kappa) \vec{H} = \mu_0(\vec{H} + \vec{M})$
* $\vec{H}\rightarrow$ **Inducing** magnetic field
* $\vec{M}\rightarrow$ **Induced** magnetism

### The anomalous field
<div class="row">
<div class="column">

* Measured field: $\vec{B} = \vec{B}_0 + \vec{B}_a$
* The total field anomaly:    $ \Delta B = |\vec{B}| - |\vec{B}_0|$
* If  $|\vec{B}_a|  <<  |\vec{B}_0|$  then:
    * The total field anomaly $\Delta B$ is the projection of the anomalous field onto the direction of the inducing field.
      $$|\vec{B}_a| \approx \vec{B}_a \cdot \hat{B}_0$$
