### Secondary Potentials!
**NOTE UPFRONT:** $V$ is the same as $\phi$. It is another common name for electric potential.

The charges on the boundaries are just charges, so Coulomb's law (lecture 12) gives their potential:

$$V_{sec}(\vec{r}) = \frac{1}{4\pi\varepsilon_0}\sum_i \frac{q_i}{|\vec{r} - \vec{r}_i|}$$

What we measure is the total:
$$V_{total} = V_{primary} + V_{secondary}$$
$$\Delta V = V_M - V_N$$

### How does that change $\rho_{app}$?
1. $\rho_{app} = \frac{\Delta V}{I}\frac{1}{G}$, and we chose $I$ and $G$, so **$\rho_{app}$ follows $\Delta V$**
2. $\Delta V$ is the "elevation drop" in potential from M to N
3. Boundary charges reshape the hillside: **+ charge makes a hill, − charge makes a pit**

| Body | Near M (A side) | Near N (B side) | Drop M to N | $\rho_{app}$ |
|---|---|---|---|---|
| Resistor | + hill | − pit | bigger | **up** |
| Conductor | − pit | + hill | smaller | **down** |

## What does the signal actually look like?

The transmitter is not *actually* a static current... it is actually an alternating square waveform with a very long on-time (reaches steady state).
* About 2 seconds of "on time"
* Potential measured towards the end of the on time
* Average alternating signed measurements to cancel out static voltages
* Repeat many times to average out noise (stacking!)
