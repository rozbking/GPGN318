## Induced Polarization (IP)
* Occurs when ionic charges accumulate within materials under an applied voltage $\rightarrow$ Generates a secondary potential
* Not an instantaneous process!
* Occurs in fluid-filled pore-spaces



### Two types:

1) Membrane polarization: Ions accumulate at pore throat:  

2) Electrode polarization: Ions accumulate at metals  


### Two scales, two effects

**Large scale: the rock body (DC resistivity)**
* Charges build up *instantly* on boundaries between rock units
* Control where current flows and the potential we measure
* We report the **bulk resistivity** at steady state (end of the on-time)

**Small scale: grains and pores (IP)**
* Ions pile up *slowly* on metal grains and in clay-lined pore throats
* Partly block current at the grain scale, so the bulk rock slowly becomes slightly **more resistive** during the on-time
* After switch-off, the ions relax and the voltage **decays**: that's the IP signal

**Same survey, two measurements:**
* On-time (steady state) → bulk resistivity, $\rho$
* Off-time (decay) → chargeability, $\eta$

### Impact on Voltage Measurements (again)

* Charge accumulation on boundaries (change in ρ)
    * $\rightarrow$ "Instantaneous" change in potential
* Charge accumulation due to IP
    * $\rightarrow$ Non-instantaneous change in potential
    * $\rightarrow$ Reaches a saturation point
    * $\rightarrow$ Measurable voltage during off-time



### Impact on Voltage Measurements (and again!)
<div class="row">
<div class="column">

1) Transmitter turns on → voltage jumps instantly to $V_\sigma$ (set by the resistivity $\rho$)
2) Ions slowly accumulate → voltage creeps up: $V_{on}(t) = V_\sigma + V_s\left(1 - e^{-t/\tau}\right)$
3) Ions saturate → steady voltage: $V_{m} = V_\sigma + V_s$
4) Transmitter turns off → the resistive part $V_\sigma$ disappears instantly, leaving $V_s$
5) IP voltage decays during the off-time: $V_{off}(t) = V_s\, e^{-t/\tau}$

$\tau$ = how fast the ions build up and relax (the time constant)
</div>

## Chargeability

* Strength of material’s IP signature represented by chargeability, i.e. its ability to be charged by an external voltage.
    
    * Really an empirical description of a "mostly" understood effect
    * Understood: the mechanisms (ions pile up at metal grains and clay-lined pores) and the trends (more sulfide → more IP; bigger grains → slower decay)
    * Not fully: no single theory predicts chargeability from rock composition, real rocks mix mechanisms, and measured values depend on the instrument settings
    * → Use IP to compare zones ("much more chargeable than background"), not to measure exact sulfide content


Two types:
1) Intrinsic Chargeability (over-voltage/DC voltage, usually in mV/V) $$\eta = \frac{V_s} {V_m}$$
2) Integrated Chargeability $$ d_{IP} = \frac{1}{V_m}\int_{t_1}^{t_2} V_s(t)dt$$






### Chargeability in Frequency Domain
* Anything time-dependent can also be measured as a function of frequency:
<div class="row">
<div class="column">

* Measure an impedance
$$ Z(\omega) = \frac{\Delta V(\omega)}{I(\omega)}$$
* $\rightarrow$ Compute resistivity
$$ \rho(\omega) = \frac{Z(\omega) A }{L}$$
* $\rightarrow$ Fit Cole-Cole model:
$$\rho(\omega) = \rho_0\left(1-\eta \left(1 - \frac{1}{1 + (i\omega \tau)^C}\right)\right)$$

</div><div class="column">


</div>
</div>


### Chargeability of Rocks:

* Some rocks are chargeable (sulfides, volcanic tuffs, clays)
* More aren’t (igneous, sandstones, limestones etc…)



### Impacts on Chargeability
* Abundance of sulfide mineralization
* Porewater salinity (# ions)
* Clay content
* Tortuosity (how twisty the pore paths are)
* Chargeability is *often* correlated with conductivity (clays, sulfides), but **not always**
    * e.g. a shale can be very conductive but barely chargeable
    * That's why IP adds new information

## Time-dependence means better gear than simply DC
* **Timing:** sample the decay in windows starting milliseconds after switch-off (Tx and Rx synchronized)
* **Clean switch-off:** a sharp current cutoff so it doesn't blur the early decay
* **Sensitive receiver:** IP voltages are small (mV) compared with the DC voltage (V) → more stacking
* **Non-polarizing potential electrodes:** steel stakes make their own IP; use porous pots for M and N
* **Cable layout:** keep Tx and Rx wires apart to reduce EM coupling

### DCIP Field Data
<div class="row">
<div class="column">

**DC Data**
* Measure during on-time
    * DC Voltage $\Delta V_m$
    * $\rho_{app} = \frac{\Delta V}{I G}$
</div><div class="column">

**IP Data**
* Measure during off-time
* Integrate over curve (like an apparent integrated chargeability)
$$ d_{IP} = \frac{1}{V_m}\int_{t_1}^{t_2} V_s(t)dt$$
* Plot on pseudo-section
    * (geometry automatically accounted for)
* Can collect both during same survey!!!

    * → so IP data are a small change in the DC data
